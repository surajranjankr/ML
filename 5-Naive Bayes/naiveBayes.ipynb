### **Naive Bayes Machine Learning Algorithms Implementation**

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
X, y = load_iris(return_X_y=True)
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
## Since continuous values, using gaussian naive bayes
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

Has some numerical and categorical values

Categorical values try to convert into ones and zeros

Try to predict anything, e.g., smoker(binary classification) or time

Take output as time and predict whether it is dinner or lunch

Try it, convert these features into one-hot encoded or label encoded

Said it should be converted into one-hot encoded, think why

One‑Hot Encoding avoids that by creating binary columns for each category:

sex_Female, sex_Male

smoker_Yes, smoker_No  
Each column is independent, so the model learns separate probabilities for each category without assuming order.

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['time'])
y = df['time']
```

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train.head()
```

```python
ct = ColumnTransformer(transformers=[('onehot', OneHotEncoder(drop='first'), [2, 3, 4])], remainder='passthrough')
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
X_train = ct.fit_transform(X_train)
```

```python
X_test = ct.transform(X_test)
```

```python
X_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
```

```python
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
