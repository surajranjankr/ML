```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
print(iris['DESCR'])
```

```python
print(iris['target'])
```

```python
## Independent feature
X = pd.DataFrame(iris['data'], columns=['sepal length in cm', 'sepal width in cm', 'petal length', 'petal width'])
```

```python
## Dependent feature
y = iris['target']
```

```python
## Train and test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

In Decision Trees, we don't require standardization or normalization.

It is because decision tree will directly apply the algorithm

```python
## Apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
treeClassifier = DecisionTreeClassifier()
```

```python
treeClassifier.fit(X_train, y_train)
```

```python
## Visualize the Decision Tree
from sklearn import tree
plt.figure(figsize=(15, 10))
tree.plot_tree(treeClassifier, filled=True)
```

```python
y_pred = treeClassifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
print(classification_report(y_test, y_pred))
```

When we create this kind of decision tree, it leads to overfitting because we are splitting the decision tree completely to it's depth

### Preventing Overfitting
1. Post Prunning
2. Pre Prunning

### **Post Prunning**
// Always good for small dataset

After watching image, we can cut the tree at max_depth=2 i.e., further split after this is not required

```python
## Apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
treeClassi = DecisionTreeClassifier(max_depth=2)

treeClassi.fit(X_train, y_train)

## Visualize the Decision Tree
from sklearn import tree
plt.figure(figsize=(15, 10))
tree.plot_tree(treeClassi, filled=True)
```

This model is more than enough to classify

```python
y_pred_classi = treeClassi.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
cm = confusion_matrix(y_test, y_pred_classi)
print(cm)
print(classification_report(y_test, y_pred_classi))
```

### **Pre Prunning**
// Usually, pre prunning is best because decision tree has many parameters

We will select best parameters using hyperparameter tuning or randomizedsearchCV

> ### Decision Tree Classifier Preprunning and Hyperparameter Tuning

```python
params = {
    'criterion' : ['gini', 'entropy', 'log_loss'],
    'splitter' : ['best', 'random'],
    'max_depth' : [1, 2, 3, 4, 5, 6],
    'max_features' : ['auto', 'sqrt', 'log2']
}
```

```python
params
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
treemodel = DecisionTreeClassifier()
```

```python
grid = GridSearchCV(treemodel, param_grid=params, cv=5, scoring='accuracy')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train, y_train)
```

```python
grid.best_params_, grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```
