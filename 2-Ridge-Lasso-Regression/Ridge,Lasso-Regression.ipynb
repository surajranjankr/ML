# **Algerian Forest Fires Dataset**

Data set information:

The dataset include 244 instances that regroup a data of two regions of Algeria, namely Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.

122 instances for each region.

The period from June 2012 to September 2012. The dataset includes 11 attributes and 1 output attribute (class). The 244 instances have been classified into fire(138 classes) and not fire (106 classes) classes.

## Attribute Information

1. **Date**: (DD/MM/YYYY) Day, month ('June' to 'September'), year (2012) — Weather data observations  
2. **Temp**: Temperature at noon (maximum) in Celsius degrees — Range: 22 to 42  
3. **RH**: Relative Humidity (%) — Range: 21 to 90  
4. **Ws**: Wind speed (km/h) — Range: 6 to 29  
5. **Rain**: Total rainfall per day (mm) — Range: 0 to 16.8  
6. **Fine Fuel Moisture Code (FFMC)**: Index from the FWI system — Range: 28.6 to 92.5  
7. **Duff Moisture Code (DMC)**: Index from the FWI system — Range: 1.1 to 65.9  
8. **Drought Code (DC)**: Index from the FWI system — Range: 7 to 220.4  
9. **Initial Spread Index (ISI)**: Index from the FWI system — Range: 0 to 18.5  
10. **Buildup Index (BUI)**: Index from the FWI system — Range: 1.1 to 68  
11. **Fire Weather Index (FWI)**: Index — Range: 0 to 31.1  
12. **Classes**: Two categories — *Fire* and *Not Fire*


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv", header=1)  # header = 1 takes head of dataset skipping the 1st row
dataset.head()
```

```python
dataset.info()
```

## Data Cleaning

```python
## Missing Values
dataset[dataset.isnull().any(axis=1)]
```

Dataset Region Split

The dataset is converted into two sets based on **Region** starting from the 122nd index.  
We can create a new column to represent the region.

1. **Bejaia Region Dataset**  
2. **Sidi‑Bel Abbes Region Dataset**

**Add a new column named `Region`** to label each record accordingly.


```python
dataset.loc[:122, "Region"] = 0
dataset.loc[122:, "Region"] = 1
df = dataset
```

```python
df.info()
```

```python
df[["Region"]] = df[["Region"]].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Removing the null values
df = df.dropna().reset_index(drop=True)  # After dropping, gap is created, so index is reset so that data is whole
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
## Remove the 122nd row
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns      # There is space in column section
```

```python
## Fix spaces in columns names
df.columns = df.columns.str.strip()
df.columns
```

```python
df.info()
```

##### Change the required columns as integer data type

```python
df.columns
```

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']] = df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

### Changing the other columns to float data type

```python
objects = [features for features in df.columns if df[features].dtypes == 'O']
```

```python
objects
```

```python
for i in objects:
    if i != 'Classes':
        df[i] = df[i].astype(float)
```

Forest Fire Observation Summary

It’s observed that August and September had the most number of forest fires for both regions.  
From the above plot of months, we can understand a few things:

- Most of the fires happened in August.  
- Very high fire activity occurred during only three months — June, July, and August.  
- September recorded fewer fires compared to the previous months.
